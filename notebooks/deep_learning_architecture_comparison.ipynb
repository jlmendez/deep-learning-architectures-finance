# Deep Learning Architectures for Finance — Comparison Notebook
Portfolio edition derived from the original architecture notebook. It separates tabular MLP experiments from recurrent sequence models and the Transformer-style implementation in `src/attention_transformer.py`.


```python
from sklearn.model_selection import train_test_split
from src.sequence_data import transaction_data, financial_sequences
from src.mlp import build_mlp
from src.recurrent import build_lstm, build_gru
from src.experiments import fit_and_score
```

```python
X, y = transaction_data()
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
mlp_result = fit_and_score(build_mlp(X.shape[1]), Xtr, ytr, Xte, yte)
mlp_result
```

```python
Xs, ys = financial_sequences()
Xs_tr, Xs_te, ys_tr, ys_te = train_test_split(Xs, ys, test_size=.25, stratify=ys, random_state=42)
lstm = fit_and_score(build_lstm(Xs.shape[1], Xs.shape[2]), Xs_tr, ys_tr, Xs_te, ys_te)
gru = fit_and_score(build_gru(Xs.shape[1], Xs.shape[2]), Xs_tr, ys_tr, Xs_te, ys_te)
lstm, gru
```

## Architecture decision
MLP is appropriate for fixed-width tabular features. LSTM/GRU add temporal memory for ordered sequences. Attention/Transformer layers become attractive when long-range interactions and parallel sequence processing matter.
